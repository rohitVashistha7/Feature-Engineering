# 1.	What is a parameter?
  
  -> In machine learning, a parameter is a variable that is learned from the training data during the training process. Parameters are used to define the model and are adjusted to minimize the error (or loss) between the predicted outputs and the actual targets (labels).

There are two main types of parameters:

- Model Parameters: These are the internal variables that the machine learning algorithm learns from the training data. For example:

  -  In a linear regression model, the parameters are the coefficients (weights) for the features and the intercept.
  - In a neural network, the parameters are the weights and biases associated with each neuron in the layers.
- Hyperparameters: These are parameters that are set before training the model and are not learned from the data. They control the learning process, such as:

  - The learning rate
  - The number of layers in a neural network
  - The number of trees in a random forest
  - Regularization strength

#6. How do we handle categorical variables in Machine Learning? What are the common techniques?

  -> To use categorical data in machine learning models, it must be converted into numerical values. Many machine learning algorithms cannot directly handle categorical variables. Here are common techniques for handling categorical data:

  - One-Hot Encoding: This involves creating a binary vector for each category, where each element of the vector represents the presence or absence of the category. Each category becomes a column with ‘1’ for the presence of that category and ‘0’ for others.
  - Label Encoding: This assigns a unique numerical value to each category in a categorical variable, with the order of the values based on the order of the categories.
  - Frequency Encoding: This replaces each category in a categorical variable with its frequency (or count) in the dataset.

  - Target Encoding: This replaces each category in a categorical variable with the mean (or other aggregation) of the target variable for that category. The goal of target encoding is to capture the relationship between the categorical variable and the target variable to improve the predictive performance of the machine learning model.
  - Ordinal Encoding: Assigns a unique integer to each label based on some ordering.
  - Binary Encoding: In binary encoding, categories are converted into binary code.

#14. How can you find correlation between variables in Python?
-> You can find the correlation between variables in Python using libraries like NumPy, SciPy, and pandas

NumPy: You can use the np.corrcoef() function to calculate the correlation coefficient between two arrays. This function returns a correlation matrix.

      import numpy as np
      var1 = np.random.randint(0, 10, 50)
      var2 = var1 + np.random.normal(0, 10, 50)
      correlation_matrix = np.corrcoef(var1, var2)
      correlation = correlation_matrix[0, 1] #to get the correlation coefficient between the two variables [1]
      print(correlation)


SciPy: The pearsonr() function from scipy.stats returns the Pearson correlation coefficient and the p-value.

    from scipy.stats.stats import pearsonr
    correlation, p_value = pearsonr(var1, var2)
    print(correlation, p_value)

Pandas: For calculating the correlation between several variables in a Pandas DataFrame, you can use the .corr() function. This returns a correlation matrix.

    import pandas as pd
    import numpy as np
    data = pd.DataFrame(np.random.randint(0, 10, size=(5, 3)), columns=['A', 'B', 'C'])
    correlation_matrix = data.corr()
    print(correlation_matrix)

To calculate the correlation between two specific variables in a DataFrame, you can specify the columns

    correlation = data['A'].corr(data['B'])
    print(correlation)


#15. What is causation? Explain difference between correlation and causation with an example.

  -> Causation indicates that one event, process, state, or object (the cause) contributes to the production of another event, process, state, or object (the effect), where the cause is at least partly responsible for the effect, and the effect is at least partly dependent on the cause. In simpler terms, causation is the action of causing something to happen or exist. It's the relationship between an act and the consequences it produces. In law, causation refers to the relationship of cause and effect between one event or action and the result

  Difference between Correlation and Causation:

Correlation describes the extent to which two variables have a statistical association, meaning they tend to change together. Causation, on the other hand, means that one event directly causes another. Just because two things are correlated does not necessarily mean one causes the other.

Example to illustrate the difference:

Correlation: Ice cream sales and crime rates might be positively correlated, meaning that as ice cream sales increase, so does crime.
Causation: It would be incorrect to conclude that ice cream sales cause crime. A more likely explanation is that both ice cream sales and crime rates increase during warmer months due to a third factor (seasonality). The warm weather is a common cause for both, but one does not directly cause the other.

#16. What is an Optimizer? What are different types of optimizers? Explain each with an example

-> An optimizer is a tool, algorithm, or strategy designed to enhance or maximize the performance of an underlying system or process. It adjusts various parameters to reach optimal outcomes across fields like engineering, finance, and manufacturing. In machine learning, optimizers are algorithms used to find the optimal set of parameters for a model during training. These algorithms iteratively adjust the weights and biases in the model until they converge on a minimum loss value.

Types of Optimizers:
1. Stochastic Gradient Descent (SGD)
Description: SGD is one of the most commonly used optimizers. It updates the model's parameters after calculating the gradient of the loss function for each individual training example. This makes it computationally efficient, especially for large datasets.

2. Mini-batch Gradient Descent
Description: This is a hybrid approach that updates the model's parameters based on a small batch of training examples (instead of just one as in SGD or all the data as in batch gradient descent).

3. Momentum Description: Momentum improves the convergence of SGD by adding a velocity term that accumulates the gradients of past steps. This helps the optimizer "build up speed" and navigate areas where gradients are small.

#21. What is feature scaling? How does it help in Machine Learning?

-> Feature scaling transforms numerical features to a common scale. It helps ML models by:
- Speeding up convergence: Gradient descent works better.

#22.  How do we perform scaling in Python?

-> To perform feature scaling in Python, you can use the sklearn.preprocessing
module, which provides various scalers.

Example:

Using Min-Max Scaling (Normalization):

Scales features to a range between 0 and 1
Uses the formula: X_scaled = (X - X_min) / (X_max - X_min)
